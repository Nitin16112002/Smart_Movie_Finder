```python
import numpy as np
import pandas as pd
import warnings
warnings.simplefilter("ignore")
```

```python
movies_data=pd.read_csv('anime.csv')
movies_data.head(3)
```

```python
movies_data.reset_index(inplace=True)
movies_data.head(2)
```

```python
movies_data.shape
```

```python
movies_data.info()
```

```python
movies_data.isnull().sum()
```

```python
# selecting the relevant features for recommendation

selected_features = ['genre','type','rating']
print(selected_features)
```

```python
# replacing the null valuess with null string

for features in selected_features:
    movies_data[features]=movies_data[features].fillna('')
```

```python
# combining all the 5 selected features

combined_features = movies_data['genre']+' '+movies_data['type']
```

```python
print(combined_features)
```

### Convert the Text Data into Vectors

```python
from sklearn.feature_extraction.text import TfidfVectorizer
vectorizer=TfidfVectorizer()
feature_vectors=vectorizer.fit_transform(combined_features)
print(feature_vectors)
```

### Cosine Similarity

```python
from sklearn.metrics.pairwise import cosine_similarity
similarity=cosine_similarity(feature_vectors)
print(similarity)
```

```python
print(similarity.shape)
```

```python
# creating a list with all the movie names given in the dataset

list_of_all_titles=movies_data['name'].tolist()
print(list_of_all_titles)
```

```python
# getting the movie name from the user

anime_name = input(' Enter your favourite anime name : ')
```

```python
# finding the close match for the movie name given by the user
import difflib
find_close_match=difflib.get_close_matches(anime_name,list_of_all_titles)
print(find_close_match)
```

```python
close_match=find_close_match[0]
print(close_match)
```

```python
# finding the index of the movie with title

index_of_the_movie=movies_data[movies_data.name==close_match]['index'].values[0]
print(index_of_the_movie)
```

```python
# getting a list of similar movies

similarity_score=list(enumerate(similarity[index_of_the_movie]))
print(similarity_score)
```

```python
# sorting the movies based on their similarity score

sorted_similar_movies=sorted(similarity_score,key= lambda x:x[1],reverse=True)
print(sorted_similar_movies)
```

```python
# print the name of similar movies based on the index

print('Movies suggested for you : \n')

i = 1

for movie in sorted_similar_movies:
  index = movie[0]
  title_from_index = movies_data[movies_data.index==index]['name'].values[0]
  if (i<11):
    print(i, '.',title_from_index)
    i+=1
```

# Smart Anime Finder

```python
anime_name = input(' Enter your favourite anime name : ')

list_of_all_titles = movies_data['name'].tolist()

find_close_match = difflib.get_close_matches(anime_name, list_of_all_titles)

close_match = find_close_match[0]

index_of_the_movie = movies_data[movies_data.name == close_match]['index'].values[0]

similarity_score = list(enumerate(similarity[index_of_the_movie]))

sorted_similar_movies = sorted(similarity_score, key = lambda x:x[1], reverse = True) 

print('Animes suggested for you : \n')

i = 1

for movie in sorted_similar_movies:
  index = movie[0]
  title_from_index = movies_data[movies_data.index==index]['name'].values[0]
  if (i<11):
    print(i, '.',title_from_index)
    i+=1
```
